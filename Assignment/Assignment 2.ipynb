## Problem 1 — Sign & parity
Ask for an integer and print whether it is:
- Positive, Negative or Zero
- Even or Odd

```python
# TODO: Classify the integer by sign and parity.
# Tip: use % 2 to test even/odd.
# ENTER INTEGER PLACE SPOT

num=int(input("ENTER INTEGER:"))

#POSITIVE
if num > 0:
    print ("Positive")

#NEGATIVE
elif num < 0:
    print ("Negative")

#NEITHER
else:
    print ("Zero")

#EVEN
if num %2==0:
    print ("Even")

#ODD
else: 
    print ("Odd")

```

## Problem 2 
Given a float x, determine if it is in the interval $[0,1], (1,2], (2,3]$ or none of them.


```python
# TODO: Implement an if-elif-else 

x = float(input("ENTER FLOAT X:"))
#[0,1]
if x >= 0 and x <= 1:
    print("In [0,1]")
#(1,2]
elif x > 1 and x <= 2:
    print("In (1,2]")
#(2,3]
elif x > 2 and x <= 3:
    print("In (2,3]")
#NONE
else:
    print("None of them")
```

## Problem 3 — Triangle Classifier
 Ask the user for three side lengths a, b, c.
 1. Check if they can form a valid triangle (all sides have positive length, and all triangle inequalities are satisfied).
 2. If valid, classify as:
    - Equilateral (all sides equal),
    - Isosceles (two sides equal),
    - Scalene (all sides different).

3. Otherwise, print that it's not a valid triangle.

Hint: you can place an `if`-block inside another `if`-block

```python
# TODO: Implement triangle classification.
# Example interaction:
# Enter side a: 3
# Enter side b: 3
# Enter side c: 3
# Output: "Equilateral triangle"

#INPUT VALUES
a = float(input("ENTER A:"))
b = float(input("ENTER B:"))
c = float(input("ENTER C:"))

#IS IT A TRIANGLE
if a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a:
#WHAT KIND OF TRIANGLE
    if a == b == c:
        print("Equilateral triangle")
    elif a == b or b == c or a == c:
        print("Isosceles triangle")
    else:
        print("Scalene triangle")
#NOT A TRIANGLE
else:
    print("INVALID TRIANGLE")
    
```
