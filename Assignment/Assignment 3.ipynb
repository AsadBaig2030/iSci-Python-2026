## Problem 1 — Sum of first N square integers
Ask the user for `N` and compute the sum $1^2 + 2^2 + ... + N^2$ using a loop.

```python
N = int(input("Enter N: "))

total = 0

for i in range(1, N+1):
    total = total + i*i

print(total)
```

## Problem 2 — Multiplication table
Print a 10x10 multiplication table (rows 1..10, columns 1..10). Format it into aligned columns.

```python
# TODO: Implement a nested loop printing a multiplication table.
# Aesthetic Tip: use f-strings with field width (e.g., f"{value:4d}") for alignment (replace "valued" with the actual integer you want to print).
# Concretely, you can use print(f"{value:d},end="") to keep printing on the same line (the 'end=""' prevents a new line from being started)
# To go to the next line, you can use print("")

for row in range(1, 11):
    for col in range(1, 11):
        print(f"{row * col:5d}", end="")
    print("")

```

## Problem 3 — Guessing game (while loop)
Choose a secret number (hardcode or random). Repeatedly ask the user to guess until correct.
- If the guess is too low/high, print a hint.
- Count the number of guesses.

```python

# TODO: Implement a guessing game using a while loop.
# Optional: import random and use random.randint(1, 100)

secret = 27
guesses = 0

while True:
    guess = int(input("Guess the number: "))
    guesses = guesses + 1

    if guess < secret:
        print("Too low")
    elif guess > secret:
        print("Too high")
    else:
        print("Correct! It took you", guesses, "guesses.")
        break
```
