## The function $f$ and its derivative $f'$

For part (d), change these two functions.

```python
import math

def f(x):
    return math.cos(x) - x

def fprime(x):
    return -math.sin(x) - 1
```

## One step of Newton's method

```python
# One step of Newton's method: given the current guess x = x_n,
# return the next guess x_{n+1}.
def newton_step(x):
    # REPLACE the x after "return" by your formula from part (b),
    # written using f(x) and fprime(x).
    return x - f(x)/fprime(x)
```

## Run Newton's method

```python
x = 1.0          # starting guess x_0
num_steps = 6    # number of Newton steps to take

print(" n                   x_n        f(x_n)")
for n in range(num_steps + 1):
    if n > 0:
        x = newton_step(x)
    print(f"{n:2d}  {x:20.15f}  {f(x):12.3e}")

# Only relevant while f(x) = x**2 - 2:
print("For comparison, math.sqrt(2) =", math.sqrt(2))
```
